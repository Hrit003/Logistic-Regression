**Definition:**

Simple Linear Regression finds the best-fitting straight line (called the regression line) through a set of data points, which predicts the value of Y based on X.

The equation of the line is:

Y = a + bX

Where:

Y → Dependent variable (the value we want to predict)

X → Independent variable (the predictor)

a → Intercept (value of Y when X = 0)

b → Slope (shows how much Y changes for a one-unit change in X)


**Purpose of Simple Linear Regression:**

1. Prediction:
To predict the value of one variable (Y) based on another variable (X).
Example: Predicting house price (Y) based on its size (X).


2. Understanding relationships:
To measure and understand how strongly X and Y are related.
Example: Studying how study hours (X) affect exam scores (Y).


3. Trend estimation:
To find trends or patterns in data (e.g., sales over time).

##Answer 2

The key assumptions of Simple Linear Regression (SLR) ensure that the model produces reliable and accurate results.
There are five main assumptions, explained below;

1. Linearity:

The relationship between the independent variable (X) and the dependent variable (Y) should be linear.

That means, changes in X lead to proportional changes in Y.


2. Independence of Errors:

The residuals (errors) should be independent of each other.

In other words, the error for one observation should not depend on the error for another.


3. Normality of Errors:

The residuals (errors) should be normally distributed.

This is important for accurate hypothesis testing and confidence intervals.

##Answer 3

The mathematical equation for a simple linear regression model is:

y= A + Bx + C

y is Dependent variable; the variable we are trying to predict or explain (also called the response or output).

x is Independent variable; the variable used to make predictions (also called the predictor, regressor, or input).

A is Intercept; the value of y when x=0; it represents the baseline or starting point of the model.

B is Slope coefficient; the amount by which y changes for a one-unit increase x; it measures the strength and direction of the relationship.

C is Error term (residual); The difference between the observed value and the predicted value; it accounts for randomness or factors not captured by the model.

##Answer 7

Here are three common evaluation metrics used for regression models, along with brief descriptions:

1. **Mean Absolute Error (MAE)**

* Definition: The average of the absolute differences between actual and predicted values.

* Description: Measures how far predictions are, on average, from the true values — simple and interpretable, but treats all errors equally.

2. **Mean Squared Error (MSE)**

* Definition: The average of the squared differences between actual and predicted values.

* Description: Gives more weight to large errors (because of squaring), making it sensitive to outliers. Useful when large mistakes are particularly undesirable.

3. **R-squared (Coefficient of Determination)**

* Definition: Indicates how well the model explains the variability of the target variable.

* Description: Ranges from 0 to 1; higher values mean a better fit. square of R= 1 represents perfect predictions, while square of R = 0 means the model performs no better than simply using the mean.

##Answer 8

The R-squared (R²) metric — also called the coefficient of determination — measures how well a regression model explains the variability of the dependent (target) variable.

* Purpose:

The purpose of R-squared is to show how much of the variation in the output variable (y) can be explained by the input variables (x) in the model.

* R² =1; The model perfectly predicts all data points.

* R² =0; The model does no better than simply predicting the mean of y.

* R² <0; The model performs worse than the mean prediction (possible in some models).

```python
##Answer 9

# Import necessary libraries
from sklearn.linear_model import LinearRegression
import numpy as np

# Example data
# X needs to be a 2D array (n_samples, n_features)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 4, 5, 4, 5])

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Print the slope (coefficient) and intercept
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```
